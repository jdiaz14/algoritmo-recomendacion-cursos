Algoritmo básico de recomendación de cursos

Objetivo

Construir un algoritmo básico de recomendación que presente dos cursos al azar, permita al usuario elegir uno de ellos y registre las decisiones tomadas.

Al finalizar, el algoritmo mostrará cuáles fueron los cursos más elegidos.

```python
import random

cursos = [
    "Python",
    "Marketing Digital",
    "Inteligencia Artificial",
    "Excel",
    "Programación Web",
    "Análisis de Datos"
]

votos = {curso: 0 for curso in cursos}

print("Cursos disponibles:")
for curso in cursos:
    print("-", curso)
```

```python
numero_de_comparaciones = 10

for i in range(numero_de_comparaciones):

    opcion1, opcion2 = random.sample(cursos, 2)

    print(f"\nComparación {i + 1}")
    print(f"1. {opcion1}")
    print(f"2. {opcion2}")

    eleccion = input("¿Cuál curso prefieres? Escribe 1 o 2: ")

    while eleccion not in ["1", "2"]:
        eleccion = input("Respuesta inválida. Escribe 1 o 2: ")

    if eleccion == "1":
        votos[opcion1] += 1
    else:
        votos[opcion2] += 1
```

```python
print("\n" + "=" * 40)
print("RESULTADOS DE LA RECOMENDACIÓN")
print("=" * 40)

resultados = sorted(votos.items(), key=lambda x: x[1], reverse=True)

for posicion, (curso, cantidad) in enumerate(resultados, start=1):
    print(f"{posicion}. {curso}: {cantidad} votos")

print("\nCurso más recomendado:")
print(resultados[0][0])
```

## ¿Cómo funciona el algoritmo?

El algoritmo utiliza una lista de cursos y selecciona aleatoriamente dos opciones
en cada comparación. El usuario debe elegir cuál de los dos cursos prefiere.

Cada elección se registra como un voto para el curso seleccionado. Este proceso
se repite 10 veces.

Finalmente, los votos se organizan de mayor a menor para identificar cuáles
cursos fueron los más elegidos.

De esta manera, el algoritmo permite obtener una recomendación basada en las
preferencias registradas durante las comparaciones.
